```python
# np.random.exponential(max(90 - 22, 0.0)) + 45

# Plot histogram of the generated exponential value
# Last bin should be 90+
import matplotlib.pyplot as plt



# Also plot histogram of time_avg_start column from the dataset
time_avg_start_values = understat_stats['time_avg_start'].tolist()
# Take zeros out
time_avg_start_values = [v for v in time_avg_start_values if v != 0]

# values = [np.random.exponential(max(v - 45, 0.0)) + 45 for v in time_avg_start_values for _ in range(1000)]
# values = [min(v, 90) for v in values]

values = [max(90 - np.random.poisson(max(90 - v, 0.0)), 45) for v in time_avg_start_values for _ in range(1000)]
values = [min(v, 90) for v in values]

# Plot as probability distribution
# Give them both the same bin width
bin_width = 1
bins = range(0, 91, bin_width) 

plt.hist(time_avg_start_values, bins=bins, alpha=0.5, label='time_avg_start', density=True)
plt.hist(values, bins=bins, alpha=0.5, label='generated_values', density=True)
plt.legend()
plt.show()
```
